# 02d · CellFM embeddings

**Environment:** `envs/cellfm.yml` (this foundation model has heavy, specific dependencies: run this notebook in its own environment).
**Output:** one `.h5ad` per dataset whose `X` is the frozen embedding (1536-d) and whose `obs` carries `cell_id`, `cell_type`, `master_split`.
The backbone is frozen: no fine-tuning, no gradients.

Get the model: clone `https://github.com/biomed-AI/CellFM` (PyTorch port + MindSpore, which is only used to read the checkpoint) and place the 80M checkpoint in `pretrained model/`.
**Caution:** frozen CellFM embeddings may be less informative than the other experts; if so, report it as a limitation. The 800M-parameter model needs a large GPU.

> These extraction functions are ported from the authors' working notebooks and need the model's own code and checkpoint.
> They are not run in the repository's tests. Try them on the smallest dataset first and check the shapes at the end.

```python
from pathlib import Path
from heterosc.extraction.cellfm import extract_cellfm_cells

DATA = Path("data")                       # from notebook 01
OUT = Path("outputs/embeddings")          # shared layout: {dataset}/{expert}/{dataset}_{expert}_with_embeddings.h5ad
DATASETS = ["Myeloid"]                    # start small; then add Lung, Liver, Pancreas, Skin
CELLFM_REPO = Path("checkpoints/CellFM-torch")
```

## Part A · annotation datasets

```python
for ds in DATASETS:
    extract_cellfm_cells(
        DATA / "annotation" / "processed" / f"{ds}_ready_master_split.h5ad",
        OUT / ds / "cellfm" / f"{ds}_cellfm_with_embeddings.h5ad",
        cellfm_repo=CELLFM_REPO)
```

## Part B · cell lines for drug response
Uses `data/drug_response/cellline_expression.h5ad` (CCLE expression, 697 genes). Output layout used by notebook 04:
`outputs/drug_response_embeddings/{expert}/cellline_{expert}_embeddings.h5ad`.

```python
extract_cellfm_cells(DATA / "drug_response" / "cellline_expression.h5ad",
                     "outputs/drug_response_embeddings/cellfm/cellline_cellfm_embeddings.h5ad",
                     cellfm_repo=CELLFM_REPO, extra_obs=())
```
